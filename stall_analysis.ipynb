#Gikomba Market Stall Performance Analysis
####This a fresh produce stall run by Wanjiku which I am going to find out how the business is doing,whether its growing,what she should restock and where she should focus.

##Data

```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]   # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg
stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

##Task 1
####In this task am going to use basic built in functions like sum(),max(), for computations and if/else for making decisions.

```python
#performing the computations
total_sales = sum(daily_sales)

avg_sale = total_sales/len(daily_sales)


highest_sale = max(daily_sales)

lowest_sale = min(daily_sales)

#identifying the day name and week number of best and worst days
best_day = days[daily_sales.index(highest_sale)]

worst_day = days[daily_sales.index(lowest_sale)]

best_day_number = daily_sales.index(highest_sale)+1

worst_day_number = daily_sales.index(lowest_sale)+1

#condition for determining week numbers
if best_day_number <= 7:
  best_day_week="1"
else:
  best_day_week="2"

if worst_day_number <= 7:
  worst_day_week = "1"
else:
  worst_day_week = "2"

```

##Task 2
###Used basic built in functions in calculations,used splicing and tuple packing and unpacking

```python
def week_totals(sales):
  #Calculating totals for each week and the week on week growth percentage
  week1_total = sum(sales[:7])
  week2_total = sum(sales[7:])
  week_on_week_growth_pct = ((week2_total - week1_total) / week1_total) * 100
  return week1_total, week2_total, week_on_week_growth_pct

# Calling the function outside of its definition

week1_total, week2_total, week_on_week_growth_pct = week_totals(daily_sales)
```

##Task 3
###In this section i have used For loop,conditional statements and dictionary counting.

```python
def classify_day(amount, avg_sale):
  #classifying days
  if amount > avg_sale * 1.2:
    return "Strong"
  elif amount < avg_sale * 0.8:
    return "Weak"
  else:
    return "Normal"

# Initializing counts outside of the function
counts = {
    "Strong": 0,
    "Normal": 0,
    "Weak": 0
}

# Looping over daily_sales outside of the function
for amount in daily_sales:
  result = classify_day(amount, avg_sale)
  counts[result] += 1

```

##Task 4
### The main cocept here is the use of list slicing.

```python
week_days=daily_sales[0:5] + daily_sales[7:12]
weekends = daily_sales[5:7] + daily_sales[12:14]
weekdays_avg = sum(week_days) / len(week_days)
weekends_avg =sum(weekends) / len(weekends)
pct_diff=(weekdays_avg-weekends_avg)/weekends_avg * 100




```

##Task 5
###In this i used for loop and conditional statement.

```python
def stock_report(stock):
  #Dtermining which items to restock
  to_reorder = []
  for items in stock:
    if stock[items]["qty"] <= stock[items]["reorder"]:
      to_reorder.append(items)
  return to_reorder

to_reorder = stock_report(stock)

# Calculating and print gross margins for all items
for items in stock:
  cost = stock[items]["cost"]
  price = stock[items]["price"]
  gross_profit = price - cost
  gross_mgn_pct = (gross_profit / price) * 100







```
